# PCA - Principal Component Analysis
Dimensionality Reduction Technique

```python
# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
# TODO: Replace with your dataset path
dataset_path = 'your_dataset.csv'
df = pd.read_csv(dataset_path)

# Display first few rows
print(df.head())
print("\nDataset shape:", df.shape)
print("\nColumn names:", df.columns.tolist())
```

```python
# Prepare the data
# TODO: If you have a target column, separate it (optional)
# For PCA, we typically work with features only

# Option 1: Use all columns
X = df

# Option 2: Drop target column if exists
# X = df.drop('target_column', axis=1)
# y = df['target_column']  # Save for visualization later

print("Features shape:", X.shape)
print("Number of features:", X.shape[1])
```

```python
# Standardize the features (IMPORTANT for PCA)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Data standardized successfully")
print("Scaled data shape:", X_scaled.shape)
```

```python
# Apply PCA to see explained variance for all components
pca_full = PCA()
pca_full.fit(X_scaled)

# Plot explained variance ratio
explained_variance = pca_full.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance)

plt.figure(figsize=(12, 5))

# Plot 1: Individual explained variance
plt.subplot(1, 2, 1)
plt.bar(range(1, len(explained_variance) + 1), explained_variance, alpha=0.7)
plt.xlabel('Principal Component')
plt.ylabel('Explained Variance Ratio')
plt.title('Explained Variance by Each Component')
plt.grid(axis='y', alpha=0.3)

# Plot 2: Cumulative explained variance
plt.subplot(1, 2, 2)
plt.plot(range(1, len(cumulative_variance) + 1), cumulative_variance, marker='o', linestyle='-')
plt.xlabel('Number of Components')
plt.ylabel('Cumulative Explained Variance')
plt.title('Cumulative Explained Variance')
plt.grid(True, alpha=0.3)
plt.axhline(y=0.95, color='r', linestyle='--', label='95% Variance')
plt.legend()

plt.tight_layout()
plt.show()

print("\nExplained Variance Ratio (first 5 components):")
for i in range(min(5, len(explained_variance))):
    print(f"PC{i+1}: {explained_variance[i]:.4f} ({explained_variance[i]*100:.2f}%)")
```

```python
# Apply PCA with reduced components
# TODO: Adjust n_components (number of principal components to keep)
n_components = 2  # Common choice: 2 or 3 for visualization

pca = PCA(n_components=n_components)
X_pca = pca.fit_transform(X_scaled)

print(f"PCA applied with {n_components} components")
print(f"Original shape: {X_scaled.shape}")
print(f"Reduced shape: {X_pca.shape}")
print(f"\nTotal variance explained: {sum(pca.explained_variance_ratio_):.4f} ({sum(pca.explained_variance_ratio_)*100:.2f}%)")
```

```python
# Create DataFrame with principal components
pca_df = pd.DataFrame(
    data=X_pca,
    columns=[f'PC{i+1}' for i in range(n_components)]
)

print("\nPCA DataFrame (first 10 rows):")
print(pca_df.head(10))

# If you have target variable, add it back
# pca_df['Target'] = y.values
```

```python
# Visualize PCA results (2D)
if n_components >= 2:
    plt.figure(figsize=(10, 6))
    plt.scatter(X_pca[:, 0], X_pca[:, 1], alpha=0.6, c='blue')
    
    # If you have target variable, color by target:
    # plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y, cmap='viridis', alpha=0.6)
    # plt.colorbar(label='Target')
    
    plt.xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.2f}% variance)')
    plt.ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.2f}% variance)')
    plt.title('PCA - 2D Visualization')
    plt.grid(True, alpha=0.3)
    plt.show()
```

```python
# Component loadings (contribution of each original feature)
loadings = pd.DataFrame(
    pca.components_.T,
    columns=[f'PC{i+1}' for i in range(n_components)],
    index=X.columns
)

print("\nPCA Component Loadings:")
print(loadings)

# Visualize loadings for PC1
plt.figure(figsize=(10, 6))
loadings['PC1'].sort_values().plot(kind='barh')
plt.xlabel('Loading Value')
plt.ylabel('Feature')
plt.title('Feature Contributions to PC1')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()
```
