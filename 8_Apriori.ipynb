# Apriori Algorithm - Implementation Template
Association Rule Mining

```python
# Import necessary libraries
import pandas as pd
import numpy as np
from mlxtend.frequent_patterns import apriori, association_rules
from mlxtend.preprocessing import TransactionEncoder
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
# TODO: Replace with your dataset path
# Note: Dataset should be in transaction format (one-hot encoded or list of transactions)
dataset_path = 'your_dataset.csv'
df = pd.read_csv(dataset_path)

# Display first few rows
print(df.head())
print("\nDataset shape:", df.shape)
```

```python
# OPTION 1: If data is already one-hot encoded (1s and 0s)
# Use this directly
# df_encoded = df

# OPTION 2: If data is in list/transaction format
# Example: [['milk', 'bread'], ['milk', 'bread', 'butter'], ['milk', 'butter']]
# Uncomment below if you need to convert:

# transactions = []  # Your list of transactions
# te = TransactionEncoder()
# te_ary = te.fit(transactions).transform(transactions)
# df_encoded = pd.DataFrame(te_ary, columns=te.columns_)

# For this template, assuming data is already encoded
df_encoded = df

print("Encoded data shape:", df_encoded.shape)
print("\nFirst few rows:")
print(df_encoded.head())
```

```python
# Apply Apriori algorithm
# TODO: Adjust min_support (threshold for frequent itemsets)
min_support = 0.01  # Minimum support (1%)

frequent_itemsets = apriori(df_encoded, min_support=min_support, use_colnames=True)

print(f"Apriori algorithm applied with min_support={min_support}")
print(f"\nNumber of frequent itemsets found: {len(frequent_itemsets)}")
print("\nTop 10 frequent itemsets:")
print(frequent_itemsets.head(10))
```

```python
# Generate association rules
# TODO: Adjust metric and min_threshold
# metric options: 'support', 'confidence', 'lift'
metric = 'confidence'
min_threshold = 0.5  # Minimum confidence (50%)

rules = association_rules(frequent_itemsets, metric=metric, min_threshold=min_threshold)

print(f"\nAssociation rules generated")
print(f"Metric: {metric}, Min threshold: {min_threshold}")
print(f"Number of rules found: {len(rules)}")
```

```python
# Display top association rules
if len(rules) > 0:
    print("\nTop 10 Association Rules:")
    print("="*80)
    
    # Sort by lift (or confidence)
    rules_sorted = rules.sort_values('lift', ascending=False)
    
    print(rules_sorted[['antecedents', 'consequents', 'support', 'confidence', 'lift']].head(10))
else:
    print("No rules found. Try lowering the thresholds.")
```

```python
# Analyze the rules
if len(rules) > 0:
    print("\nRule Statistics:")
    print("="*50)
    print(f"Average Support: {rules['support'].mean():.4f}")
    print(f"Average Confidence: {rules['confidence'].mean():.4f}")
    print(f"Average Lift: {rules['lift'].mean():.4f}")
    print(f"\nMax Lift: {rules['lift'].max():.4f}")
    print(f"Max Confidence: {rules['confidence'].max():.4f}")
```

```python
# Visualize: Support vs Confidence
if len(rules) > 0:
    plt.figure(figsize=(10, 6))
    scatter = plt.scatter(rules['support'], rules['confidence'], 
                         c=rules['lift'], cmap='viridis', alpha=0.6, s=100)
    plt.xlabel('Support')
    plt.ylabel('Confidence')
    plt.title('Association Rules: Support vs Confidence (colored by Lift)')
    plt.colorbar(scatter, label='Lift')
    plt.grid(True)
    plt.show()
```

```python
# Display specific strong rules
if len(rules) > 0:
    print("\nStrong Rules (High Lift):")
    print("="*80)
    strong_rules = rules[rules['lift'] > 1.0].sort_values('lift', ascending=False)
    
    for idx, row in strong_rules.head(5).iterrows():
        print(f"\nRule {idx}:")
        print(f"  If {row['antecedents']} → Then {row['consequents']}")
        print(f"  Support: {row['support']:.4f}, Confidence: {row['confidence']:.4f}, Lift: {row['lift']:.4f}")
```
