Simple implementation for end semester practical

```python
# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error
import matplotlib.pyplot as plt
```

```python
# Load the dataset
# TODO: Replace with your dataset path
dataset_path = 'your_dataset.csv'
df = pd.read_csv(dataset_path)

# Display first few rows
print(df.head())
print("\nDataset shape:", df.shape)
```

```python
# Prepare the data
# TODO: Replace 'target_column' with your target variable name
# TODO: Select feature columns (X should have ONE column for simple linear regression)
X = df[['feature_column']]  # Independent variable
y = df['target_column']      # Dependent variable

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
# Split the data into training and testing sets
# TODO: Adjust test_size if needed (0.2 = 20% test, 80% train)
test_size = 0.2
random_state = 42

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=random_state)

print("Training set size:", X_train.shape[0])
print("Testing set size:", X_test.shape[0])
```

```python
# Create and train the model
model = LinearRegression()
model.fit(X_train, y_train)

print("Model trained successfully!")
print("Coefficient:", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
# Make predictions
y_pred = model.predict(X_test)

print("Predictions made on test set")
print("\nFirst 5 predictions vs actual:")
comparison = pd.DataFrame({'Actual': y_test[:5].values, 'Predicted': y_pred[:5]})
print(comparison)
```

```python
# Evaluate the model
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Model Evaluation Metrics:")
print("="*40)
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
print(f"Mean Absolute Error (MAE): {mae:.4f}")
print(f"R-squared (R2) Score: {r2:.4f}")
```

```python
# Visualize the results
plt.figure(figsize=(10, 6))
plt.scatter(X_test, y_test, color='blue', label='Actual', alpha=0.6)
plt.scatter(X_test, y_pred, color='red', label='Predicted', alpha=0.6)
plt.plot(X_test, y_pred, color='black', linewidth=2, label='Regression Line')
plt.xlabel('Feature')
plt.ylabel('Target')
plt.title('Linear Regression: Actual vs Predicted')
plt.legend()
plt.grid(True)
plt.show()
```
