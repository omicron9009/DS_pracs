# Logistic Regression - Implementation Template
Binary/Multi-class Classification

```python
# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
# TODO: Replace with your dataset path
dataset_path = 'your_dataset.csv'
df = pd.read_csv(dataset_path)

# Display first few rows
print(df.head())
print("\nDataset shape:", df.shape)
print("\nColumn names:", df.columns.tolist())
```

```python
# Prepare the data
# TODO: Replace 'target_column' with your target variable name
X = df.drop('target_column', axis=1)  # Features
y = df['target_column']                # Target (class labels)

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nClass distribution:")
print(y.value_counts())
```

```python
# Split the data into training and testing sets
# TODO: Adjust test_size if needed
test_size = 0.2
random_state = 42

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=random_state)

print("Training set size:", X_train.shape[0])
print("Testing set size:", X_test.shape[0])
```

```python
# Create and train Logistic Regression model
# TODO: Adjust hyperparameters if needed
# solver options: 'lbfgs', 'liblinear', 'newton-cg', 'sag', 'saga'
max_iter = 100  # Maximum iterations
solver = 'lbfgs'

model = LogisticRegression(max_iter=max_iter, solver=solver, random_state=random_state)
model.fit(X_train, y_train)

print(f"Logistic Regression model trained")
print(f"Max iterations: {max_iter}")
print(f"Solver: {solver}")
```

```python
# Make predictions
y_pred = model.predict(X_test)

print("Predictions made on test set")
print("\nFirst 10 predictions vs actual:")
comparison = pd.DataFrame({'Actual': y_test[:10].values, 'Predicted': y_pred[:10]})
print(comparison)
```

```python
# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)

print("Model Evaluation:")
print("="*40)
print(f"Accuracy: {accuracy:.4f}")
print(f"Accuracy Percentage: {accuracy*100:.2f}%")
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
# Get prediction probabilities
y_pred_proba = model.predict_proba(X_test)

print("\nPrediction Probabilities (first 5 samples):")
print(y_pred_proba[:5])
print(f"\nClasses: {model.classes_}")
```

```python
# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
print("\nConfusion Matrix:")
print(cm)

# Visualize Confusion Matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=True)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Logistic Regression - Confusion Matrix')
plt.show()
```

```python
# Model coefficients (feature importance)
if len(model.classes_) == 2:  # Binary classification
    coefficients = pd.DataFrame({
        'Feature': X.columns,
        'Coefficient': model.coef_[0]
    }).sort_values('Coefficient', ascending=False)
    
    print("\nFeature Coefficients:")
    print(coefficients)
    
    # Visualize
    plt.figure(figsize=(10, 6))
    plt.barh(coefficients['Feature'], coefficients['Coefficient'])
    plt.xlabel('Coefficient Value')
    plt.ylabel('Feature')
    plt.title('Logistic Regression - Feature Coefficients')
    plt.gca().invert_yaxis()
    plt.show()
```
