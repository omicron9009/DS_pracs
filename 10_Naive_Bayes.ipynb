# Naive Bayes - Implementation Template
Probabilistic Classification Algorithm

```python
# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
# TODO: Replace with your dataset path
dataset_path = 'your_dataset.csv'
df = pd.read_csv(dataset_path)

# Display first few rows
print(df.head())
print("\nDataset shape:", df.shape)
print("\nColumn names:", df.columns.tolist())
```

```python
# Prepare the data
# TODO: Replace 'target_column' with your target variable name
X = df.drop('target_column', axis=1)  # Features
y = df['target_column']                # Target (class labels)

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nClass distribution:")
print(y.value_counts())
```

```python
# Split the data into training and testing sets
# TODO: Adjust test_size if needed
test_size = 0.2
random_state = 42

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=random_state)

print("Training set size:", X_train.shape[0])
print("Testing set size:", X_test.shape[0])
```

```python
# Create and train Naive Bayes model
# GaussianNB is used for continuous features
# Note: For discrete features, use MultinomialNB or BernoulliNB

model = GaussianNB()
model.fit(X_train, y_train)

print("Naive Bayes (Gaussian) model trained successfully")
```

```python
# Make predictions
y_pred = model.predict(X_test)

print("Predictions made on test set")
print("\nFirst 10 predictions vs actual:")
comparison = pd.DataFrame({'Actual': y_test[:10].values, 'Predicted': y_pred[:10]})
print(comparison)
```

```python
# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)

print("Model Evaluation:")
print("="*40)
print(f"Accuracy: {accuracy:.4f}")
print(f"Accuracy Percentage: {accuracy*100:.2f}%")
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
# Get prediction probabilities
y_pred_proba = model.predict_proba(X_test)

print("\nPrediction Probabilities (first 5 samples):")
print(y_pred_proba[:5])
print(f"\nClasses: {model.classes_}")
```

```python
# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
print("\nConfusion Matrix:")
print(cm)

# Visualize Confusion Matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Greens', cbar=True)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Naive Bayes - Confusion Matrix')
plt.show()
```

```python
# Class prior probabilities
print("\nClass Prior Probabilities:")
print("="*40)
for i, cls in enumerate(model.classes_):
    print(f"Class {cls}: {model.class_prior_[i]:.4f}")
```

```python
# Analyze predictions by class
print("\nPrediction Distribution:")
print("="*40)
pred_dist = pd.Series(y_pred).value_counts().sort_index()
print(pred_dist)

# Visualize
plt.figure(figsize=(10, 6))
pred_dist.plot(kind='bar', color='skyblue')
plt.xlabel('Class')
plt.ylabel('Count')
plt.title('Naive Bayes - Prediction Distribution')
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3)
plt.show()
```
