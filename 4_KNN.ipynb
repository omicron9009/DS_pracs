# K-Nearest Neighbors (KNN) - Implementation Template
Classification using KNN

```python
# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
# TODO: Replace with your dataset path
dataset_path = 'your_dataset.csv'
df = pd.read_csv(dataset_path)

# Display first few rows
print(df.head())
print("\nDataset shape:", df.shape)
print("\nColumn names:", df.columns.tolist())
```

```python
# Prepare the data
# TODO: Replace with your feature column names
# TODO: Replace 'target_column' with your target variable name
X = df.drop('target_column', axis=1)  # Features
y = df['target_column']                # Target (class labels)

print("Features shape:", X.shape)
print("Target shape:", y.shape)
print("\nClass distribution:")
print(y.value_counts())
```

```python
# Split the data into training and testing sets
# TODO: Adjust test_size if needed
test_size = 0.2
random_state = 42

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=test_size, random_state=random_state)

print("Training set size:", X_train.shape[0])
print("Testing set size:", X_test.shape[0])
```

```python
# Create and train the KNN model
# TODO: Adjust k (number of neighbors) if needed
k = 5  # Number of neighbors

model = KNeighborsClassifier(n_neighbors=k)
model.fit(X_train, y_train)

print(f"KNN model trained with k={k} neighbors")
```

```python
# Make predictions
y_pred = model.predict(X_test)

print("Predictions made on test set")
print("\nFirst 10 predictions vs actual:")
comparison = pd.DataFrame({'Actual': y_test[:10].values, 'Predicted': y_pred[:10]})
print(comparison)
```

```python
# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)

print("Model Evaluation:")
print("="*40)
print(f"Accuracy: {accuracy:.4f}")
print(f"Accuracy Percentage: {accuracy*100:.2f}%")
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)

# Visualize Confusion Matrix
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Greens', cbar=True)
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title(f'KNN (k={k}) - Confusion Matrix')
plt.show()
```

```python
# Optional: Test different k values to find the best one
k_values = range(1, 21)
accuracies = []

for k_val in k_values:
    knn_temp = KNeighborsClassifier(n_neighbors=k_val)
    knn_temp.fit(X_train, y_train)
    y_pred_temp = knn_temp.predict(X_test)
    accuracies.append(accuracy_score(y_test, y_pred_temp))

# Plot accuracy for different k values
plt.figure(figsize=(10, 6))
plt.plot(k_values, accuracies, marker='o', linestyle='-', color='blue')
plt.xlabel('Number of Neighbors (k)')
plt.ylabel('Accuracy')
plt.title('KNN: Accuracy vs k value')
plt.grid(True)
plt.show()

best_k = k_values[accuracies.index(max(accuracies))]
print(f"\nBest k value: {best_k} with accuracy: {max(accuracies):.4f}")
```
