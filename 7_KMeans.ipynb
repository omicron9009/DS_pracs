# K-Means Clustering - Implementation Template
Unsupervised clustering algorithm

```python
# Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load the dataset
# TODO: Replace with your dataset path
dataset_path = 'your_dataset.csv'
df = pd.read_csv(dataset_path)

# Display first few rows
print(df.head())
print("\nDataset shape:", df.shape)
print("\nColumn names:", df.columns.tolist())
```

```python
# Prepare the data
# TODO: Select feature columns for clustering
# Note: K-Means doesn't use target variable (unsupervised)
feature_columns = ['feature1', 'feature2']  # Adjust as needed
X = df[feature_columns]

print("Features shape:", X.shape)
print("\nFirst few rows of features:")
print(X.head())
```

```python
# Standardize the features (important for K-Means)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Data standardized")
print("Scaled data shape:", X_scaled.shape)
```

```python
# Find optimal k using Elbow Method
inertias = []
k_range = range(1, 11)

for k in k_range:
    kmeans_temp = KMeans(n_clusters=k, random_state=42)
    kmeans_temp.fit(X_scaled)
    inertias.append(kmeans_temp.inertia_)

# Plot Elbow Curve
plt.figure(figsize=(10, 6))
plt.plot(k_range, inertias, marker='o', linestyle='-', color='blue')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia')
plt.title('Elbow Method - Finding Optimal k')
plt.grid(True)
plt.show()
```

```python
# Create and train K-Means model
# TODO: Adjust k (number of clusters) based on elbow plot
k = 3  # Number of clusters
random_state = 42

model = KMeans(n_clusters=k, random_state=random_state)
model.fit(X_scaled)

print(f"K-Means model trained with k={k} clusters")
print(f"\nCluster centers:\n{model.cluster_centers_}")
```

```python
# Get cluster labels
labels = model.labels_

# Add cluster labels to original dataframe
df['Cluster'] = labels

print("Cluster assignments:")
print(df['Cluster'].value_counts().sort_index())
print("\nFirst 10 rows with cluster labels:")
print(df.head(10))
```

```python
# Visualize the clusters (works best with 2 features)
plt.figure(figsize=(10, 6))
scatter = plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=labels, cmap='viridis', alpha=0.6)
plt.scatter(model.cluster_centers_[:, 0], model.cluster_centers_[:, 1], 
            c='red', marker='X', s=200, edgecolors='black', label='Centroids')
plt.xlabel(feature_columns[0])
plt.ylabel(feature_columns[1])
plt.title(f'K-Means Clustering (k={k})')
plt.colorbar(scatter, label='Cluster')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Analyze cluster statistics
print("Cluster Statistics:")
print("="*50)
for i in range(k):
    cluster_data = df[df['Cluster'] == i][feature_columns]
    print(f"\nCluster {i}:")
    print(f"Size: {len(cluster_data)}")
    print(f"Mean values:\n{cluster_data.mean()}")
```
